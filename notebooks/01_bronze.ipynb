# 01 — Bronze: ingestão das transações

Código espelhado de `RETREINO/create_delta_table.py`. A adaptação necessária é o caminho do volume e a tabela permanente `desloc.bronze.transacoes`.

```python
%sql
SELECT * FROM read_files(
  "/Volumes/desloc/bronze/transactions/transacao.csv",
  sep => ";"
)
LIMIT 5
```

```python
df = spark.read.format("csv")\
    .option("sep", ";")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("/Volumes/desloc/bronze/transactions/transacao.csv")

df.limit(5).display()
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("desloc.bronze.transacoes")
```
