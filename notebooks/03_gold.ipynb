# 03 — Gold: tabelas de EDA e ML

Espelha as seleções `tbl_ml` e `tbl_eda` de `RETREINO/00_struct_table.py`, com a origem `desloc.silver.transacoes`. As duas seleções são gravadas como tabelas permanentes no schema Gold.

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS desloc.gold")
spark.sql("""
    CREATE OR REPLACE TABLE desloc.gold.transacoes_ml AS
    SELECT
        vlr_pago,
        dia,
        dia_semana,
        mes,
        hora,
        periodo,
        distancia_lalo,
        latitude_origem,
        longitude_origem,
        latitude_destino,
        longitude_destino
    FROM desloc.silver.transacoes
""")
```

```python
spark.sql("""
    CREATE OR REPLACE TABLE desloc.gold.transacoes_eda AS
    SELECT
        vlr_pago,
        dia,
        dia_semana,
        mes,
        hora,
        periodo,
        distancia_lalo,
        duracao
    FROM desloc.silver.transacoes
""")
```

## Conferência
As tabelas Gold substituem as views temporárias `tbl_ml` e `tbl_eda` do material do professor.

```python
display(spark.sql("SELECT * FROM desloc.gold.transacoes_ml LIMIT 5"))
```

```python
display(spark.sql("SELECT * FROM desloc.gold.transacoes_eda LIMIT 5"))
```
